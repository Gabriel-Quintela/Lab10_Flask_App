```python
from app import app
from models import db
from sqlalchemy import inspect, text

with app.app_context():

    # Create database and tables if they do not exist
    db.create_all()

    # Check existing User table columns
    columns = {
        column["name"]
        for column in inspect(db.engine).get_columns("users")
    }

    # Add display_name if it does not exist
    with db.engine.begin() as connection:

        if "display_name" not in columns:
            connection.execute(text(
                "ALTER TABLE users ADD COLUMN display_name VARCHAR(100)"
            ))

        if "image_filename" not in columns:
            connection.execute(text(
                "ALTER TABLE users ADD COLUMN image_filename VARCHAR(255)"
            ))

    print("Profile database columns are ready!")
```
