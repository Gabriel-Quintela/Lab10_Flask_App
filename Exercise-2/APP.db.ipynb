```python
from app import app
from models import db

with app.app_context():
    inspector = db.inspect(db.engine)
    columns = [column["name"] for column in inspector.get_columns("users")]

    if "role" not in columns:
        db.session.execute(
            db.text("ALTER TABLE users ADD COLUMN role VARCHAR(20) NOT NULL DEFAULT 'viewer'")
        )
        db.session.commit()

    print("User role column is ready!")
```
